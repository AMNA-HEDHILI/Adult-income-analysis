```python
#find missing values
print(df.isnull().sum(),df.isnull())
```

```python
#replace ? with NaN
print(df.isin(['?']).sum())
import numpy as np
df = df.replace('?', np.nan)
```

```python
# Visualize missing values
sns.heatmap(df.isnull())
plt.show()
```

```python
# Check percentage of missing values
per_missing = df.isnull().sum() * 100 / len(df)
print(per_missing)
```

```python
#  Drop all rows with Missing Values (Notice: NO inplace=True when assigning to df!)
df = df.dropna(how='any')
print("Shape after dropping missing values:", df.shape)
```

```python
#  Find and drop duplicates
print('Are there any duplicated values in data:', df.duplicated().any())
print('Number of duplicates:', df.duplicated().sum())
```

```python
df = df.drop_duplicates()
print("Final shape after dropping duplicates:", df.shape)
```

```python
#get overall statistics about dataframe
print(df.describe())
#drop columns
df =df.drop(columns=['educational-num','capital-gain','capital-loss'])
print(df)
```

```python
#What Is The Distribution of Age Column?
print(df['age'].describe())
# Plot distribution for the 'age' column
sns.histplot(df['age'], kde=True, color='blue')

plt.title('Distribution of Age')
plt.xlabel('Age')
plt.ylabel('Count')
plt.show()
```

```python
#  distribution for the 'workclass' column
print(df['workclass'].value_counts(normalize=True) * 100)
# Create a bar plot for a categorical column
sns.countplot(data=df, x='workclass')
# Create a bar plot for a categorical column
sns.countplot(data=df, x='workclass')

# Optional: Rotate x-axis labels if the text is long and overlaps
plt.xticks(rotation=45, ha='right')

plt.title('Distribution of workclass Levels')
plt.show()
```

```python
#Find Total Number of Persons Having Age Between 17 To 48 (Inclusive) Using Between Method
print('people with age between 17 to 48',df['age'].between(17,48).sum())
```

```python
#persons with bachelors and master degree
print('Unique education:',df['education'].unique())
print('people with bachelor and Masters degree',len(df.loc[df['education'].isin(['Bachelors', 'Masters'])]))
```

```python
#Bivariate analysis simply means analyzing two variables at the same time to see if there is a relationship, pattern, or correlation between them.

#1. salary vs age
sns.boxplot(data=df, x='income', y='age')
plt.title('Age Distribution by Income Level')
plt.show()
#2.See income distribution across different education levels
sns.countplot(data=df, x='education', hue='income')
plt.xticks(rotation=45, ha='right')
plt.title('Income Level by Education')
plt.show()
```

```python
#replace salary values with 0 and 1
print('unique salary',df['income'].unique())
df['income'] = df['income'].replace({'<=50K': 0, '>50K': 1})
print(df['income'] )
```

```python
#which workclass getting the highest salary
print(df.columns)
print(df.groupby('workclass')['income'].mean().sort_values(ascending=False).head(1) * 100)
```

```python
#which gender has better chance to get salary greater than 50k 
print(df.groupby('gender')['income'].mean() * 100)
```

```python
# Convert 'workclass' to a categorical datatype
df['workclass'] = df['workclass'].astype('category')

# Verify the change
print(df['workclass'].dtype)
```
